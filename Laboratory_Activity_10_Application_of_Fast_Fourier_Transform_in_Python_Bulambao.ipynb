1. Create a composite signal that consists the following frequencies and amplitudes:

    a. 1200 Hz, 10 units,

    b. 3500 Hz, 15, units,

    c. 5000 Hz, 8 units,

    d. 8500 Hz, 12 units,

```python
import numpy as np
import matplotlib.pyplot as plt

# Signal parameters
frequencies = [1200, 3500, 5000, 8500]  # Hz
amplitudes = [10, 15, 8, 12]            # Units

# Sampling parameters
sampling_rate = 20000  # Samples per second
duration = 1            # Second
t = np.arange(0, duration, 1 / sampling_rate)

# Create the composite signal
composite_signal = np.zeros(len(t))

for f, A in zip(frequencies, amplitudes):
    composite_signal += A * np.sin(2 * np.pi * f * t)

# Plot the composite signal
plt.figure(figsize=(12, 5))
plt.plot(t[:200], composite_signal[:200])
plt.title("Composite Signal (First 200 Samples)")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude (units)")
plt.grid(True)
plt.show()
```

2. Now, perform Discrete Fourier Transform calculation using the NumPy function, np.fft.fft.


```python
import numpy as np
import matplotlib.pyplot as plt

# Signal parameters
frequencies = [1200, 3500, 5000, 8500]
amplitudes = [10, 15, 8, 12]

sampling_rate = 20000
duration = 1
t = np.arange(0, duration, 1 / sampling_rate)

# Generate composite signal
signal = np.zeros(len(t))

for f, A in zip(frequencies, amplitudes):
    signal += A * np.sin(2 * np.pi * f * t)

# Calculate DFT using NumPy FFT
N = len(signal)
dft_result = np.fft.fft(signal)

# Calculate magnitude
magnitude = np.abs(dft_result)

# Plot DFT magnitude
plt.figure(figsize=(12, 5))
plt.plot(magnitude)
plt.title("Discrete Fourier Transform (DFT)")
plt.xlabel("Frequency Bin")
plt.ylabel("Magnitude")
plt.grid(True)
plt.show()
```

3. Plot its Frequency Domain and compare the output Frequency Domain signal with Laboratory No. 9.

```python

import numpy as np
import matplotlib.pyplot as plt

# Signal parameters
frequencies = [1200, 3500, 5000, 8500]
amplitudes = [10, 15, 8, 12]

sampling_rate = 20000
duration = 1
t = np.arange(0, duration, 1 / sampling_rate)

# Generate composite signal
signal = np.zeros(len(t))

for f, A in zip(frequencies, amplitudes):
    signal += A * np.sin(2 * np.pi * f * t)

# Calculate FFT
N = len(signal)
fft_result = np.fft.fft(signal)

# Obtain positive frequency values and normalized magnitudes
frequency_axis = np.fft.rfftfreq(N, d=1 / sampling_rate)
magnitude = np.abs(np.fft.rfft(signal)) / N

# Double the positive-frequency amplitudes, excluding DC and Nyquist
magnitude[1:-1] *= 2

# Plot the frequency domain
plt.figure(figsize=(12, 5))
plt.plot(frequency_axis, magnitude)
plt.title("Frequency Domain of the Composite Signal")
plt.xlabel("Frequency (Hz)")
plt.ylabel("Amplitude (units)")
plt.xlim(0, sampling_rate / 2)
plt.grid(True)
plt.show()
```
