1. Create a composite signal consisting of six (6) different sinusoidal waves with frequencies of 20 Hz, 80 Hz, 200 Hz, 500 Hz, 800 Hz and 2000 Hz, with amplitudes of 10, 9, 11, 14, 7, 12 for each frequencies respectively, and phases of , , π, , , , respectively. Plot its first 1 second.

```python
import numpy as np
import matplotlib.pyplot as plt

# Sampling parameters
fs = 10000  # Sampling frequency in Hz
t = np.linspace(0, 1, fs, endpoint=False)

# Frequencies (Hz)
frequencies = [20, 80, 200, 500, 800, 2000]

# Amplitudes
amplitudes = [10, 9, 11, 14, 7, 12]

# Phases in radians
phases = [0, 0, np.pi, 0, 0, 0]

# Generate the composite signal
composite_signal = np.zeros_like(t)

for f, A, phase in zip(frequencies, amplitudes, phases):
    composite_signal += A * np.sin(2 * np.pi * f * t + phase)

# Plot the first 1 second
plt.figure(figsize=(14, 5))
plt.plot(t, composite_signal, linewidth=0.7)

plt.title("Composite Signal of Six Sinusoidal Waves")
plt.xlabel("Time (seconds)")
plt.ylabel("Amplitude")
plt.xlim(0, 1)
plt.grid(True)
plt.tight_layout()
plt.show()
```

2. Identify the sampling rate for proper sampling. Then, plot the sampled signal with n index for 1 second.

```python
import numpy as np
import matplotlib.pyplot as plt

# Sampling parameters
fs = 10000  # Sampling rate in Hz
Ts = 1 / fs  # Sampling period
n = np.arange(fs)  # Sample index for 1 second
t = n * Ts  # Discrete-time sampling instants

# Frequencies (Hz)
frequencies = [20, 80, 200, 500, 800, 2000]

# Amplitudes
amplitudes = [10, 9, 11, 14, 7, 12]

# Phases in radians
phases = [0, 0, np.pi, 0, 0, 0]

# Generate the sampled composite signal
x_n = np.zeros(len(n))

for f, A, phase in zip(frequencies, amplitudes, phases):
    x_n += A * np.sin(2 * np.pi * f * n / fs + phase)

# Plot sampled signal using discrete n index
plt.figure(figsize=(14, 5))
plt.stem(n, x_n, linefmt='b-', markerfmt=' ', basefmt='k-')

plt.title("Sampled Composite Signal for 1 Second")
plt.xlabel("Sample Index (n)")
plt.ylabel("Amplitude")
plt.xlim(0, fs - 1)
plt.grid(True)
plt.tight_layout()
plt.show()
```

3. Now, perform Discrete Fourier Transform calculation using the basis functions. Plot its Re X[k] and Im X[k] values.

```python

import numpy as np
import matplotlib.pyplot as plt

# Sampling parameters
fs = 10000
N = fs
n = np.arange(N)
t = n / fs

# Composite signal parameters
frequencies = [20, 80, 200, 500, 800, 2000]
amplitudes = [10, 9, 11, 14, 7, 12]
phases = [0, 0, np.pi, 0, 0, 0]

# Generate sampled composite signal
x = np.zeros(N)

for f, A, phase in zip(frequencies, amplitudes, phases):
    x += A * np.sin(2 * np.pi * f * t + phase)

# DFT calculation using basis functions
k = np.arange(N)

# Calculate real and imaginary parts using the DFT basis
X_real = np.zeros(N)
X_imag = np.zeros(N)

for ki in k:
    angle = 2 * np.pi * ki * n / N

    X_real[ki] = np.sum(x * np.cos(angle))
    X_imag[ki] = -np.sum(x * np.sin(angle))

# Plot Re{X[k]}
plt.figure(figsize=(14, 5))
plt.stem(k, X_real, linefmt='b-', markerfmt=' ', basefmt='k-')
plt.title('Real Part of DFT: Re{X[k]}')
plt.xlabel('Frequency Index k')
plt.ylabel('Re{X[k]}')
plt.xlim(0, N - 1)
plt.grid(True)
plt.tight_layout()
plt.show()

# Plot Im{X[k]}
plt.figure(figsize=(14, 5))
plt.stem(k, X_imag, linefmt='r-', markerfmt=' ', basefmt='k-')
plt.title('Imaginary Part of DFT: Im{X[k]}')
plt.xlabel('Frequency Index k')
plt.ylabel('Im{X[k]}')
plt.xlim(0, N - 1)
plt.grid(True)
plt.tight_layout()
plt.show()
```

4. Compute the Mag X[k] and Phase X[k] of the signal as well.

```python

import numpy as np
import matplotlib.pyplot as plt

# Sampling parameters
fs = 10000
N = fs
n = np.arange(N)
t = n / fs

# Composite signal parameters
frequencies = [20, 80, 200, 500, 800, 2000]
amplitudes = [10, 9, 11, 14, 7, 12]
phases = [0, 0, np.pi, 0, 0, 0]

# Generate composite signal
x = np.zeros(N)

for f, A, phase in zip(frequencies, amplitudes, phases):
    x += A * np.sin(2 * np.pi * f * t + phase)

# Compute DFT
X = np.fft.fft(x)

# Calculate magnitude and phase
mag_X = np.abs(X)
phase_X = np.angle(X)

# Frequency index
k = np.arange(N)

# Plot magnitude spectrum
plt.figure(figsize=(14, 5))
plt.stem(k, mag_X, linefmt='b-', markerfmt=' ', basefmt='k-')
plt.title('Magnitude Spectrum |X[k]|')
plt.xlabel('Frequency Index k')
plt.ylabel('Magnitude |X[k]|')
plt.xlim(0, N - 1)
plt.grid(True)
plt.tight_layout()
plt.show()

# Plot phase spectrum
plt.figure(figsize=(14, 5))
plt.stem(k, phase_X, linefmt='r-', markerfmt=' ', basefmt='k-')
plt.title('Phase Spectrum ∠X[k]')
plt.xlabel('Frequency Index k')
plt.ylabel('Phase (radians)')
plt.xlim(0, N - 1)
plt.grid(True)
plt.tight_layout()
plt.show()
```
