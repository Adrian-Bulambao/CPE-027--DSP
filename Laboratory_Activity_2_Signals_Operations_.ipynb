# **3. Discussion:**


Signals constitute an important part of our daily life. Anything that carries some information is called a signal. A signal is defined as a single-valued function of one or more independent variables which contain some information. A signal is also defined as a physical quantity that varies with time, space or any other independent variable. A signal may be represented in time domain or frequency domain. Human speech is a familiar example of a signal. Electric current and voltage are also examples of signals. A signal can be a function of one or more independent variables. A signal may be a function of time, temperature, position, pressure, distance etc. If a signal depends on only one independent variable, it is called a one- dimensional signal, and if a signal depends on two independent variables, it is called a two- dimensional signal.

When we process a signal, this signal may undergo several manipulations involving the independent variable or the amplitude of the signal. The basic operations on sequences are as follows:
1. Time shifting
2. Time reversal
3. Time scaling
4. Amplitude scaling
5. Signal addition
6. Signal multiplication


The first three operations correspond to transformation in independent variable n of a signal. The last three operations correspond to transformation on amplitude of a signal.

# **OUTPUT**

1. Using NumPy libraries, create a sinusoidal signal using the following formula:<br>
x1t=Asin(2πft+∅)<br>
where A represents the amplitude of the signal, f denotes the frequency of the signal and ϕ indicates the phase of the signal. Set A = 3.2, f = 2.4, and ∅=3π4.

```python
# importing libraries
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
A = 3.2
f = 2.4
phi = (3 * np.pi) / 4
t = np.linspace(0, 2, 500)

x = A * np.sin(2 * np.pi * f * t + phi)

plt.figure(figsize=(10, 5))
plt.plot(t, x, label=f'x = {A}·sin(2π·{f}t + {phi:.2f})', color='purple', linewidth=2)

plt.axhline(0, color='black', linewidth=0.8, linestyle='--')  # Center axis line
plt.title('Graph of a Sinusoidal Equation')
plt.xlabel('Time (t) in seconds')
plt.ylabel('Displacement (x)')
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='upper right')
plt.show()
```

2. Using the created sinusoidal signal, perform time advance with 3 units. Then graph and compare the result signal to original sinusoidal signal.

```python
A = 3.2
f = 2.4
phi = (3 * np.pi) / 4
t = np.linspace(0, 2, 500)
advanced = 3

x_orig = A * np.sin(2 * np.pi * f * t + phi)
x_advance = A * np.sin(2 * np.pi * f * (t + advanced) + phi)

plt.figure(figsize=(12, 6))

plt.plot(t, x_orig, label='Original: $x(t)$', color='purple', linewidth=2.5)

plt.plot(t, x_advance, label='Time Advanced: $x(t + 3)$', color='darkorange', linewidth=2, linestyle='--')

plt.axhline(0, color='black', linewidth=0.8, linestyle='-')
plt.title('Comparison: Original vs. Time-Advanced Sinusoidal Signal', fontsize=14, fontweight='bold')
plt.xlabel('Time ($t$) in seconds', fontsize=12)
plt.ylabel('Displacement ($x$)', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='upper right', fontsize=11)
plt.xlim(0, 2)

plt.show()
```

3. Perform time reversal with the created sinusoidal signal. Then graph and compare the result signal to original sinusoidal signal.

```python
A = 3.2
f = 2.4
phi = (3 * np.pi) / 4
t = np.linspace(0, 2, 500)

x_orig = A * np.sin(2 * np.pi * f * t + phi)
x_reverse = A * np.sin(2 * np.pi * f * (-t) + phi)

plt.figure(figsize=(12, 6))

plt.plot(t, x_orig, label='Original: $x(t)$', color='purple', linewidth=2.5)

plt.plot(t, x_reverse, label='Time Reversed: $x(-t)$', color='darkorange', linewidth=2, linestyle='--')

plt.axhline(0, color='black', linewidth=0.8, linestyle='-')
plt.title('Comparison: Original vs. Time Reversed Sinusoidal Signal', fontsize=14, fontweight='bold')
plt.xlabel('Time ($t$) in seconds', fontsize=12)
plt.ylabel('Displacement ($x$)', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='upper right', fontsize=11)
plt.xlim(0, 2)  # Focus on the 2-second window

plt.show()
```

4. Perform time scaling using a=0.4 with the created sinusoidal signal. Then graph and compare the result signal to original sinusoidal signal.

```python
A1 = 3.2
A2 = 0.4
f = 2.4
phi = (3 * np.pi) / 4
t = np.linspace(0, 2, 500)

x_orig = A1 * np.sin(2 * np.pi * f * t + phi)
x_time_scaled = A * np.sin(2 * np.pi * f * (t* A2) + phi)

plt.figure(figsize=(12, 6))

plt.plot(t, x_orig, label='Original: $x(t)$', color='purple', linewidth=2.5)
plt.plot(t, x_time_scaled, label='Time-Scaled: $x({a}t)$ (Expanded)', color='darkorange', linewidth=2, linestyle='--')


plt.axhline(0, color='black', linewidth=0.8, linestyle='-')
plt.title('Comparison: Original vs. Time- Sinusoidal Signal', fontsize=14, fontweight='bold')
plt.xlabel('Time ($t$) in seconds', fontsize=12)
plt.ylabel('Displacement ($x$)', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='upper right', fontsize=11)
plt.xlim(0, 2)

plt.show()
```

5. Perform amplitude scaling using a=1/4 with the created sinusoidal signal. Then graph and compare the result signal to original sinusoidal signal.

```python
A1 = 3.2
A2 = 1/4
f = 2.4
phi = (3 * np.pi) / 4
t = np.linspace(0, 2, 500)

x_orig = A1 * np.sin(2 * np.pi * f * t + phi)
x_ampli_scaled = A2 * x_orig

plt.figure(figsize=(12, 6))

plt.plot(t, x_orig, label='Original: $x(t)$', color='purple', linewidth=2.5)
plt.plot(t, x_ampli_scaled, label='Amplitude-Scaled: $\\frac{{1}}{{4}}x(t)$ ($A=0.8$)', color='darkorange', linewidth=2, linestyle='--')


plt.axhline(0, color='black', linewidth=0.8, linestyle='-')
plt.title('Comparison: Original vs. Amplitude-Scaled Sinusoidal Signal', fontsize=14, fontweight='bold')
plt.xlabel('Time ($t$) in seconds', fontsize=12)
plt.ylabel('Displacement ($x$)', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='upper right', fontsize=11)
plt.xlim(0, 2)

plt.show()
```

6. Create a separate exponential signal using the following formula:<br>
x2t=at <br>
Graph the result signal where a = 1.8


```python
import numpy as np
import matplotlib.pyplot as plt

a = 1.8
t = np.linspace(0, 2, 500)

x2 = a ** t


plt.figure(figsize=(10, 6))
plt.plot(t, x2, label=f'$x_2(t) = {a}^t$', color='crimson', linewidth=2.5)

plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='gray', linewidth=0.8, linestyle='--')

plt.title('Graph of an Exponential Growth Signal', fontsize=14, fontweight='bold')
plt.xlabel('Time ($t$) in seconds', fontsize=12)
plt.ylabel('$x_2(t)$', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='upper left', fontsize=11)

plt.show()
```

7. Perform signal addition for x1 and x2 and plot the result.


```python
t = np.linspace(0, 2, 500)

A = 3.2
f = 2.4
phi = (3 * np.pi) / 4
x1 = A * np.sin(2 * np.pi * f * t + phi)

a = 1.8
x2 = a ** t

x_sum = x1 + x2

plt.figure(figsize=(12, 6))

plt.plot(t, x1, label='Sinusoidal: $x_1(t)$', color='purple', alpha=0.5, linestyle=':')
plt.plot(t, x2, label='Exponential: $x_2(t)$', color='crimson', alpha=0.5, linestyle='--')
plt.plot(t, x_sum, label='Added: $x_1(t) + x_2(t)$', color='darkblue', linewidth=2.5)

plt.axhline(0, color='black', linewidth=0.8, linestyle='-')
plt.title('Signal Addition: $x_1(t) + x_2(t)$', fontsize=14, fontweight='bold')
plt.xlabel('Time ($t$) in seconds', fontsize=12)
plt.ylabel('Amplitude', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='upper left', fontsize=11)
plt.xlim(0, 2)

plt.show()
```

8. Perform signal multiplication for x1 and x2. Then graph and compare the product signal to summed signal.

```python
t = np.linspace(0, 2, 500)

A = 3.2
f = 2.4
phi = (3 * np.pi) / 4
x1 = A * np.sin(2 * np.pi * f * t + phi)

a = 1.8
x2 = a ** t

x_sum = x1 + x2
x_product = x1 * x2

plt.figure(figsize=(12, 6))

plt.plot(t, x1, label='Sinusoidal: $x_1(t)$', color='purple', alpha=0.5, linestyle=':')
plt.plot(t, x2, label='Exponential: $x_2(t)$', color='crimson', alpha=0.5, linestyle='--')
plt.plot(t, x_sum, label='Added: $x_1(t) + x_2(t)$', color='darkblue', linewidth=2.5)
plt.plot(t, x_product, label='Multiplied: $x_1(t) * x_2(t)$', color='green', linewidth=2.5)

plt.axhline(0, color='black', linewidth=0.8, linestyle='-')
plt.title('Signal Multiplication: $x_1(t) * x_2(t)$', fontsize=14, fontweight='bold')
plt.xlabel('Time ($t$) in seconds', fontsize=12)
plt.ylabel('Amplitude', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='upper left', fontsize=11)
plt.xlim(0, 2)

plt.show()
```
